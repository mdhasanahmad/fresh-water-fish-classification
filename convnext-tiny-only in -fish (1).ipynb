```python

# =====================================================================
# A0 ABLATION STUDY: ConvNeXt-Tiny ONLY
# Freshwater Fish Classification | Kaggle GPU / T4
# ConvNeXt-Tiny -> Global Average Pooling -> Linear Classifier
# NO GRID GRAPH | NO GCN
# Metrics + Bootstrap 95% CI + Confusion Matrix + CSV + Model Checkpoint
# =====================================================================

import os
import glob
import copy
import random
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights
from PIL import Image, ImageFile
from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, cohen_kappa_score, matthews_corrcoef,
    confusion_matrix, classification_report
)

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True

# ======================== 1. CONFIGURATION ============================

# Change this path only if your dataset is stored elsewhere.
ROOT_DIR = "/kaggle/input/datasets/mdhasanahmad/fresh-water-dataset/Original Image"

SEED = 42
BATCH_SIZE = 32
EPOCHS = 10
LR = 1e-3
WEIGHT_DECAY = 1e-4
PATIENCE = 4
NUM_BOOT = 2000
NUM_WORKERS = 2

OUT_DIR = "/kaggle/working/A0_ConvNeXt_Tiny_Ablation"
os.makedirs(OUT_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)

print("=" * 70)
print("A0: ConvNeXt-Tiny Baseline Ablation")
print("=" * 70)
print("Device:", device)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

if not os.path.isdir(ROOT_DIR):
    raise FileNotFoundError(
        f"Dataset folder not found:\n{ROOT_DIR}\n"
        "Please set ROOT_DIR to the actual Kaggle dataset directory."
    )

# ======================== 2. DATASET ==================================

class FreshwaterFishDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform

        self.classes = sorted([
            d for d in os.listdir(root_dir)
            if os.path.isdir(os.path.join(root_dir, d))
        ])

        if len(self.classes) < 2:
            raise ValueError(
                "Expected at least two class subfolders inside ROOT_DIR."
            )

        self.class_to_idx = {
            name: i for i, name in enumerate(self.classes)
        }

        valid_ext = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
        self.samples = []

        for class_name in self.classes:
            class_dir = os.path.join(root_dir, class_name)

            for current_root, _, filenames in os.walk(class_dir):
                for filename in filenames:
                    if os.path.splitext(filename)[1].lower() in valid_ext:
                        self.samples.append((
                            os.path.join(current_root, filename),
                            self.class_to_idx[class_name]
                        ))

        self.samples.sort(key=lambda x: x[0])

        if len(self.samples) == 0:
            raise RuntimeError("No valid images found in the dataset.")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):
        path, label = self.samples[index]

        with Image.open(path) as img:
            img = img.convert("RGB")

        if self.transform is not None:
            img = self.transform(img)

        return img, label


weights = ConvNeXt_Tiny_Weights.DEFAULT
transform = weights.transforms()

dataset = FreshwaterFishDataset(
    ROOT_DIR,
    transform=transform
)

class_names = dataset.classes
num_classes = len(class_names)
labels = np.array([label for _, label in dataset.samples])
indices = np.arange(len(dataset))

print("\nClasses:", class_names)
print("Number of classes:", num_classes)
print("Total images:", len(dataset))

print("\nImages per class:")
for name in class_names:
    count = sum(1 for _, y in dataset.samples
                if y == dataset.class_to_idx[name])
    print(f"{name}: {count}")

# ======================== 3. DATA SPLIT ================================

# 64% train / 16% validation / 20% test.
# Stratified random split with a fixed seed.
train_val_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=SEED,
    stratify=labels
)

train_idx, val_idx = train_test_split(
    train_val_idx,
    test_size=0.20,
    random_state=SEED,
    stratify=labels[train_val_idx]
)

train_ds = Subset(dataset, train_idx.tolist())
val_ds = Subset(dataset, val_idx.tolist())
test_ds = Subset(dataset, test_idx.tolist())

loader_kwargs = {
    "batch_size": BATCH_SIZE,
    "num_workers": NUM_WORKERS,
    "pin_memory": torch.cuda.is_available()
}

train_loader = DataLoader(
    train_ds, shuffle=True, **loader_kwargs
)
val_loader = DataLoader(
    val_ds, shuffle=False, **loader_kwargs
)
test_loader = DataLoader(
    test_ds, shuffle=False, **loader_kwargs
)

print("\nData split:")
print(f"Training:   {len(train_ds)} ({len(train_ds)/len(dataset)*100:.1f}%)")
print(f"Validation: {len(val_ds)} ({len(val_ds)/len(dataset)*100:.1f}%)")
print(f"Testing:    {len(test_ds)} ({len(test_ds)/len(dataset)*100:.1f}%)")

# ======================== 4. A0 MODEL ==================================

class ConvNeXtTinyBaseline(nn.Module):
    """
    A0 baseline:
    Pretrained ConvNeXt-Tiny feature extractor (frozen)
    -> global average pooling
    -> linear classification head.

    No graph construction and no GCN layers.
    """
    def __init__(self, n_classes):
        super().__init__()

        backbone = convnext_tiny(weights=weights)
        self.features = backbone.features

        # Keep the backbone frozen to match the intended
        # frozen-feature-extractor experimental setting.
        for param in self.features.parameters():
            param.requires_grad = False

        self.global_pool = nn.AdaptiveAvgPool2d((1, 1))
        self.classifier = nn.Linear(768, n_classes)

    def forward(self, x):
        with torch.no_grad():
            x = self.features(x)

        x = self.global_pool(x)
        x = torch.flatten(x, start_dim=1)
        x = self.classifier(x)

        return x


model = ConvNeXtTinyBaseline(num_classes).to(device)

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.classifier.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY
)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

print("\nModel: ConvNeXt-Tiny + GAP + Linear Classifier")
print(f"Total parameters:     {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print("Graph/GCN layers: None")

# ======================== 5. EVALUATION FUNCTION ========================

@torch.no_grad()
def evaluate(model, loader):
    model.eval()

    total_loss = 0.0
    total = 0
    all_true = []
    all_pred = []
    all_prob = []

    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)

        logits = model(images)
        loss = criterion(logits, targets)
        probabilities = torch.softmax(logits, dim=1)
        predictions = probabilities.argmax(dim=1)

        batch_n = targets.size(0)
        total_loss += loss.item() * batch_n
        total += batch_n

        all_true.append(targets.cpu().numpy())
        all_pred.append(predictions.cpu().numpy())
        all_prob.append(probabilities.cpu().numpy())

    y_true = np.concatenate(all_true)
    y_pred = np.concatenate(all_pred)
    y_prob = np.concatenate(all_prob)

    return (
        total_loss / max(total, 1),
        y_true,
        y_pred,
        y_prob
    )

# ======================== 6. TRAINING ==================================

history = []
best_val_loss = float("inf")
best_state = None
patience_counter = 0

training_start = time.time()

for epoch in range(1, EPOCHS + 1):
    model.train()

    running_loss = 0.0
    correct = 0
    seen = 0

    progress = tqdm(
        train_loader,
        desc=f"Epoch {epoch}/{EPOCHS}"
    )

    for images, targets in progress:
        images = images.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        logits = model(images)
        loss = criterion(logits, targets)

        loss.backward()
        optimizer.step()

        batch_n = targets.size(0)
        running_loss += loss.item() * batch_n
        correct += (logits.argmax(dim=1) == targets).sum().item()
        seen += batch_n

        progress.set_postfix(
            loss=f"{running_loss/max(seen, 1):.4f}",
            acc=f"{100*correct/max(seen, 1):.2f}%"
        )

    train_loss = running_loss / max(seen, 1)
    train_acc = correct / max(seen, 1)

    val_loss, val_true, val_pred, val_prob = evaluate(
        model, val_loader
    )
    val_acc = accuracy_score(val_true, val_pred)

    history.append({
        "Epoch": epoch,
        "Train Loss": train_loss,
        "Train Accuracy": train_acc,
        "Validation Loss": val_loss,
        "Validation Accuracy": val_acc
    })

    print(
        f"\nEpoch {epoch}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc*100:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc*100:.2f}%"
    )

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        patience_counter = 0
    else:
        patience_counter += 1

        if patience_counter >= PATIENCE:
            print("\nEarly stopping triggered.")
            break

if best_state is None:
    raise RuntimeError("Training did not produce a valid model checkpoint.")

model.load_state_dict(best_state)

training_time = time.time() - training_start
history_df = pd.DataFrame(history)

print(f"\nTraining time: {training_time/60:.2f} minutes")

# ======================== 7. FINAL TEST EVALUATION =====================

test_loss, y_true, y_pred, y_prob = evaluate(
    model, test_loader
)

def calculate_metrics(y_true, y_pred, y_prob):
    metrics = {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro-Precision": precision_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro-Recall": recall_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro-F1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Cohen's Kappa": cohen_kappa_score(y_true, y_pred),
        "MCC": matthews_corrcoef(y_true, y_pred)
    }

    try:
        metrics["Macro AUC (OvR)"] = roc_auc_score(
            y_true,
            y_prob,
            labels=np.arange(num_classes),
            multi_class="ovr",
            average="macro"
        )
    except ValueError:
        metrics["Macro AUC (OvR)"] = np.nan

    return metrics


point_metrics = calculate_metrics(y_true, y_pred, y_prob)

# ======================== 8. BOOTSTRAP 95% CI ===========================

# Percentile bootstrap over test examples.
# These intervals quantify test-sample uncertainty; they do not
# measure variation across independent training runs.
rng = np.random.default_rng(SEED)
n_test = len(y_true)

bootstrap_values = {
    name: [] for name in point_metrics.keys()
}

for _ in tqdm(range(NUM_BOOT), desc="Bootstrap confidence intervals"):
    sampled = rng.integers(0, n_test, size=n_test)

    yt = y_true[sampled]
    yp = y_pred[sampled]
    ypb = y_prob[sampled]

    # Metrics always defined for resampled labels/predictions.
    bootstrap_values["Accuracy"].append(
        accuracy_score(yt, yp)
    )
    bootstrap_values["Macro-Precision"].append(
        precision_score(yt, yp, average="macro", zero_division=0)
    )
    bootstrap_values["Macro-Recall"].append(
        recall_score(yt, yp, average="macro", zero_division=0)
    )
    bootstrap_values["Macro-F1"].append(
        f1_score(yt, yp, average="macro", zero_division=0)
    )
    bootstrap_values["Cohen's Kappa"].append(
        cohen_kappa_score(yt, yp)
    )
    bootstrap_values["MCC"].append(
        matthews_corrcoef(yt, yp)
    )

    try:
        auc_value = roc_auc_score(
            yt,
            ypb,
            labels=np.arange(num_classes),
            multi_class="ovr",
            average="macro"
        )
        bootstrap_values["Macro AUC (OvR)"].append(auc_value)
    except ValueError:
        # Some bootstrap samples may omit one or more classes.
        pass

# Corrected results table construction:
# No conditional expression is used as a dictionary key.
results = []

for metric_name, estimate in point_metrics.items():
    samples = np.asarray(
        bootstrap_values[metric_name], dtype=float
    )
    samples = samples[np.isfinite(samples)]

    if len(samples) > 0:
        ci_lower, ci_upper = np.percentile(
            samples, [2.5, 97.5]
        )
    else:
        ci_lower, ci_upper = np.nan, np.nan

    is_percentage = metric_name in [
        "Accuracy",
        "Macro-Precision",
        "Macro-Recall",
        "Macro-F1"
    ]

    if is_percentage:
        estimate_out = estimate * 100
        lower_out = ci_lower * 100
        upper_out = ci_upper * 100
        unit = "%"
    else:
        estimate_out = estimate
        lower_out = ci_lower
        upper_out = ci_upper
        unit = "score"

    results.append({
        "ID": "A0",
        "Model": "ConvNeXt-Tiny",
        "Metric": metric_name,
        "Unit": unit,
        "Estimate": estimate_out,
        "95% CI Lower": lower_out,
        "95% CI Upper": upper_out
    })

results_df = pd.DataFrame(results)

print("\n" + "=" * 80)
print("A0: CONVNEXT-TINY TEST RESULTS")
print("=" * 80)
print(f"Test loss: {test_loss:.4f}")
print(f"Test sample count: {len(y_true)}")
print(f"Training time: {training_time/60:.2f} minutes")
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")

display(
    results_df.style.format({
        "Estimate": "{:.4f}",
        "95% CI Lower": "{:.4f}",
        "95% CI Upper": "{:.4f}"
    })
)

# ======================== 9. CONFUSION MATRIX ==========================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(num_classes)
)

fig, ax = plt.subplots(
    figsize=(max(9, num_classes * 0.65),
             max(7, num_classes * 0.55))
)

im = ax.imshow(cm, interpolation="nearest", cmap="Blues")
ax.set_title("A0: ConvNeXt-Tiny — Test Confusion Matrix")
fig.colorbar(im, ax=ax)

ax.set(
    xticks=np.arange(num_classes),
    yticks=np.arange(num_classes),
    xticklabels=class_names,
    yticklabels=class_names,
    xlabel="Predicted Label",
    ylabel="True Label"
)

plt.setp(
    ax.get_xticklabels(),
    rotation=90,
    ha="center",
    fontsize=8
)
plt.setp(ax.get_yticklabels(), fontsize=8)

threshold = cm.max() / 2 if cm.size else 0

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(
            j, i, str(cm[i, j]),
            ha="center",
            va="center",
            color="white" if cm[i, j] > threshold else "black",
            fontsize=8
        )

plt.tight_layout()
cm_fig_path = os.path.join(OUT_DIR, "A0_confusion_matrix.png")
plt.savefig(cm_fig_path, dpi=300, bbox_inches="tight")
plt.show()

# ======================== 10. TRAINING CURVES ===========================

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].plot(
    history_df["Epoch"], history_df["Train Loss"],
    marker="o", label="Training"
)
axes[0].plot(
    history_df["Epoch"], history_df["Validation Loss"],
    marker="s", label="Validation"
)
axes[0].set_title("Training and Validation Loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(
    history_df["Epoch"],
    history_df["Train Accuracy"] * 100,
    marker="o", label="Training"
)
axes[1].plot(
    history_df["Epoch"],
    history_df["Validation Accuracy"] * 100,
    marker="s", label="Validation"
)
axes[1].set_title("Training and Validation Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy (%)")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
curves_path = os.path.join(OUT_DIR, "A0_training_curves.png")
plt.savefig(curves_path, dpi=300, bbox_inches="tight")
plt.show()

# ======================== 11. CLASSIFICATION REPORT ====================

report = classification_report(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

print("\nClassification report:")
display(report_df.round(4))

# ======================== 12. SAVE ALL RESULTS ==========================

history_df.to_csv(
    os.path.join(OUT_DIR, "A0_training_history.csv"),
    index=False
)

results_df.to_csv(
    os.path.join(OUT_DIR, "A0_test_metrics_bootstrap_CI.csv"),
    index=False
)

report_df.to_csv(
    os.path.join(OUT_DIR, "A0_classification_report.csv")
)

pd.DataFrame(
    cm, index=class_names, columns=class_names
).to_csv(
    os.path.join(OUT_DIR, "A0_confusion_matrix.csv")
)

checkpoint_path = os.path.join(
    OUT_DIR, "A0_ConvNeXt_Tiny.pth"
)

torch.save({
    "model_state_dict": model.state_dict(),
    "classes": class_names,
    "seed": SEED,
    "best_validation_loss": best_val_loss,
    "test_loss": test_loss,
    "test_metrics": point_metrics,
    "total_parameters": total_params,
    "trainable_parameters": trainable_params
}, checkpoint_path)

print("\n" + "=" * 80)
print("A0 ABLATION COMPLETED SUCCESSFULLY")
print("=" * 80)
print("Saved files:")
for filename in sorted(os.listdir(OUT_DIR)):
    print(" -", os.path.join(OUT_DIR, filename))

print("\nFinal model: ConvNeXt-Tiny + Global Average Pooling + Linear Head")
print("Graph: Not used")
print("GCN: Not used")
print("Confidence intervals: Bootstrap percentile 95% CI")
```

```python
# ================================================================
# A1 ABLATION STUDY: ConvNeXt-Tiny + ONE-LAYER GCN
# Freshwater Fish Image Classification | Kaggle GPU / Tesla T4
# ConvNeXt spatial features + 7x7 grid graph + 1 GCN layer
# Metrics, 95% Bootstrap CIs, Confusion Matrix, and Saved Results
# ================================================================

import os
import random
import copy
import json
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
warnings.filterwarnings("ignore")

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    cohen_kappa_score, matthews_corrcoef, confusion_matrix,
    classification_report, roc_auc_score, average_precision_score
)
from sklearn.preprocessing import label_binarize

# -------------------- 1. CONFIGURATION ----------------------------

SEED = 42
ROOT_DIR = "/kaggle/input/datasets/mdhasanahmad/fresh-water-dataset/Original Image"
OUTPUT_DIR = "/kaggle/working/A1_ConvNeXt_Tiny_1GCN_Ablation"

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 15
PATIENCE = 4
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
GCN_HIDDEN = 256
BOOTSTRAP_ROUNDS = 2000
NUM_WORKERS = 2

os.makedirs(OUTPUT_DIR, exist_ok=True)

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

if not os.path.isdir(ROOT_DIR):
    raise FileNotFoundError(
        f"Dataset directory not found:\n{ROOT_DIR}\n"
        "Update ROOT_DIR to the actual Kaggle dataset folder."
    )

# -------------------- 2. DATASET ---------------------------------

IMG_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff")

class FreshwaterFishDataset(Dataset):
    """
    Expected directory structure:
        Original Image/
            Class_1/
                image1.jpg
            Class_2/
                image2.jpg

    Each immediate subdirectory is treated as one class.
    """

    def __init__(self, root_dir, transform=None, class_to_idx=None):
        self.root_dir = root_dir
        self.transform = transform

        if class_to_idx is None:
            classes = sorted([
                d for d in os.listdir(root_dir)
                if os.path.isdir(os.path.join(root_dir, d))
            ])
            self.class_to_idx = {name: i for i, name in enumerate(classes)}
        else:
            self.class_to_idx = class_to_idx

        self.classes = [
            name for name, idx in sorted(
                self.class_to_idx.items(), key=lambda item: item[1]
            )
        ]

        self.samples = []
        for class_name in self.classes:
            class_dir = os.path.join(root_dir, class_name)
            if not os.path.isdir(class_dir):
                continue

            for current_root, _, files in os.walk(class_dir):
                for filename in files:
                    if filename.lower().endswith(IMG_EXTENSIONS):
                        self.samples.append((
                            os.path.join(current_root, filename),
                            self.class_to_idx[class_name]
                        ))

        self.samples.sort(key=lambda x: x[0])

        if not self.samples:
            raise RuntimeError(
                f"No supported images found in {root_dir}. "
                "Check the dataset directory structure."
            )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):
        path, label = self.samples[index]
        try:
            image = Image.open(path).convert("RGB")
        except Exception as exc:
            raise RuntimeError(f"Could not read image: {path}") from exc

        if self.transform is not None:
            image = self.transform(image)

        return image, label


weights = ConvNeXt_Tiny_Weights.DEFAULT
image_transform = weights.transforms()

full_dataset = FreshwaterFishDataset(
    ROOT_DIR,
    transform=image_transform
)

class_names = full_dataset.classes
num_classes = len(class_names)
all_labels = np.array([label for _, label in full_dataset.samples])
all_indices = np.arange(len(full_dataset))

print("\nDataset:", ROOT_DIR)
print("Total images:", len(full_dataset))
print("Number of classes:", num_classes)
print("Classes:", class_names)

class_counts = pd.Series(all_labels).value_counts().sort_index()
print("\nClass distribution:")
for i, name in enumerate(class_names):
    print(f"  {name}: {int(class_counts.get(i, 0))}")

if num_classes < 2:
    raise ValueError("At least two classes are required.")

# -------------------- 3. STRATIFIED SPLIT -------------------------
# 64% training, 16% validation, 20% testing.
# The test set is held out from model fitting and early stopping.

trainval_idx, test_idx = train_test_split(
    all_indices,
    test_size=0.20,
    random_state=SEED,
    stratify=all_labels
)

train_idx, val_idx = train_test_split(
    trainval_idx,
    test_size=0.20,
    random_state=SEED,
    stratify=all_labels[trainval_idx]
)

class IndexedDataset(Dataset):
    def __init__(self, base_dataset, indices):
        self.base_dataset = base_dataset
        self.indices = list(indices)

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        return self.base_dataset[self.indices[idx]]

train_dataset = IndexedDataset(full_dataset, train_idx)
val_dataset = IndexedDataset(full_dataset, val_idx)
test_dataset = IndexedDataset(full_dataset, test_idx)

loader_generator = torch.Generator()
loader_generator.manual_seed(SEED)

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available(),
    generator=loader_generator
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)

print(
    f"\nSplit sizes: train={len(train_dataset)}, "
    f"validation={len(val_dataset)}, test={len(test_dataset)}"
)

# -------------------- 4. BUILD 7x7 GRID GRAPH ---------------------
# ConvNeXt-Tiny's final spatial feature map is typically 7 x 7
# for 224 x 224 input images. Each spatial location is a graph node.
#
# Edges connect up, down, left, right neighbors, plus self-loops.
# Symmetric normalized adjacency: D^(-1/2) (A + I) D^(-1/2).

def make_grid_adjacency(height=7, width=7):
    n = height * width
    adjacency = torch.zeros((n, n), dtype=torch.float32)

    for r in range(height):
        for c in range(width):
            i = r * width + c

            # Self-loop
            adjacency[i, i] = 1.0

            # Four-neighbor grid connections
            for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                nr, nc = r + dr, c + dc
                if 0 <= nr < height and 0 <= nc < width:
                    j = nr * width + nc
                    adjacency[i, j] = 1.0

    degree = adjacency.sum(dim=1)
    degree_inv_sqrt = torch.pow(degree.clamp(min=1.0), -0.5)
    D_inv_sqrt = torch.diag(degree_inv_sqrt)
    adjacency_norm = D_inv_sqrt @ adjacency @ D_inv_sqrt
    return adjacency_norm


# -------------------- 5. MODEL: CONVNEXT + ONE GCN ----------------

class OneLayerGCN(nn.Module):
    """
    Single graph-convolution layer:
        H = ReLU(A_norm X W + b)

    Input X has shape [batch, nodes, channels].
    """

    def __init__(self, in_channels, hidden_channels, adjacency):
        super().__init__()
        self.register_buffer("adjacency", adjacency)
        self.linear = nn.Linear(in_channels, hidden_channels)
        self.norm = nn.LayerNorm(hidden_channels)

    def forward(self, x):
        # Feature transformation: [B, N, C] -> [B, N, H]
        x = self.linear(x)

        # Neighbor aggregation using normalized grid adjacency
        x = torch.matmul(self.adjacency.unsqueeze(0), x)

        x = self.norm(x)
        x = F.relu(x)
        return x


class ConvNeXtTinyOneGCN(nn.Module):
    def __init__(self, num_classes, gcn_hidden=256, dropout=0.30):
        super().__init__()

        backbone = convnext_tiny(weights=weights)
        self.backbone = backbone

        # Freeze the pretrained ConvNeXt backbone for this ablation.
        for parameter in self.backbone.parameters():
            parameter.requires_grad = False

        # Keep the frozen backbone in evaluation mode.
        self.backbone.eval()

        # ConvNeXt-Tiny's final feature channels = 768.
        self.gcn = OneLayerGCN(
            in_channels=768,
            hidden_channels=gcn_hidden,
            adjacency=make_grid_adjacency(7, 7)
        )

        # Mean-pool graph node embeddings to form an image representation.
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(gcn_hidden, num_classes)

    def train(self, mode=True):
        # Train GCN/head, but keep frozen backbone in eval mode.
        super().train(mode)
        self.backbone.eval()
        return self

    def forward(self, images):
        # Frozen backbone: [B, 3, 224, 224] -> [B, 768, 7, 7]
        with torch.no_grad():
            feature_map = self.backbone.features(images)

        if feature_map.shape[1] != 768:
            raise RuntimeError(
                f"Expected 768 ConvNeXt channels; got {feature_map.shape[1]}"
            )

        # Support a non-7x7 map safely by adaptive pooling.
        if feature_map.shape[-2:] != (7, 7):
            feature_map = F.adaptive_avg_pool2d(feature_map, (7, 7))

        # Turn spatial feature locations into graph nodes:
        # [B, 768, 7, 7] -> [B, 49, 768]
        nodes = feature_map.flatten(2).transpose(1, 2)

        # Exactly one GCN layer
        node_embeddings = self.gcn(nodes)  # [B, 49, gcn_hidden]

        # Global mean pooling over graph nodes
        graph_embedding = node_embeddings.mean(dim=1)

        logits = self.classifier(self.dropout(graph_embedding))
        return logits


model = ConvNeXtTinyOneGCN(
    num_classes=num_classes,
    gcn_hidden=GCN_HIDDEN,
    dropout=0.30
).to(device)

trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)
total_params = sum(p.numel() for p in model.parameters())

print("\nModel: ConvNeXt-Tiny + 1-layer GCN")
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")

# -------------------- 6. OPTIMIZATION -----------------------------

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=2
)

# -------------------- 7. TRAINING / VALIDATION ---------------------

def run_epoch(model, loader, optimizer=None):
    training = optimizer is not None
    model.train(training)

    total_loss = 0.0
    total_correct = 0
    total_count = 0

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        if training:
            optimizer.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(training):
            logits = model(images)
            loss = criterion(logits, labels)

            if training:
                loss.backward()
                optimizer.step()

        batch_size = labels.size(0)
        total_loss += loss.item() * batch_size
        total_correct += (logits.argmax(dim=1) == labels).sum().item()
        total_count += batch_size

    return total_loss / max(total_count, 1), total_correct / max(total_count, 1)


history = {
    "epoch": [],
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": []
}

best_val_loss = float("inf")
best_state = None
epochs_without_improvement = 0
start_time = time.time()

print("\nStarting training...\n")

for epoch in range(1, EPOCHS + 1):
    train_loss, train_acc = run_epoch(model, train_loader, optimizer)
    val_loss, val_acc = run_epoch(model, val_loader, optimizer=None)

    scheduler.step(val_loss)

    history["epoch"].append(epoch)
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train loss {train_loss:.4f} | Train acc {train_acc*100:.2f}% | "
        f"Val loss {val_loss:.4f} | Val acc {val_acc*100:.2f}%"
    )

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        epochs_without_improvement = 0
        torch.save(
            best_state,
            os.path.join(OUTPUT_DIR, "best_model.pt")
        )
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= PATIENCE:
        print(f"Early stopping at epoch {epoch}.")
        break

if best_state is not None:
    model.load_state_dict(best_state)

training_seconds = time.time() - start_time
print(f"\nTraining time: {training_seconds:.1f} seconds")

pd.DataFrame(history).to_csv(
    os.path.join(OUTPUT_DIR, "training_history.csv"), index=False
)

# -------------------- 8. TRAINING CURVES --------------------------

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].plot(history["epoch"], history["train_loss"], marker="o",
             label="Training loss")
axes[0].plot(history["epoch"], history["val_loss"], marker="s",
             label="Validation loss")
axes[0].set_title("A1: Training and Validation Loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Cross-entropy loss")
axes[0].legend()
axes[0].grid(alpha=0.25)

axes[1].plot(history["epoch"],
             np.array(history["train_accuracy"]) * 100,
             marker="o", label="Training accuracy")
axes[1].plot(history["epoch"],
             np.array(history["val_accuracy"]) * 100,
             marker="s", label="Validation accuracy")
axes[1].set_title("A1: Training and Validation Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy (%)")
axes[1].legend()
axes[1].grid(alpha=0.25)

plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "training_curves.png"),
    dpi=300, bbox_inches="tight"
)
plt.show()

# -------------------- 9. TEST PREDICTIONS ------------------------

@torch.no_grad()
def predict_model(model, loader):
    model.eval()
    y_true, y_pred, y_prob = [], [], []

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        logits = model(images)
        probabilities = torch.softmax(logits, dim=1)

        y_true.extend(labels.numpy().tolist())
        y_pred.extend(logits.argmax(dim=1).cpu().numpy().tolist())
        y_prob.extend(probabilities.cpu().numpy().tolist())

    return (
        np.asarray(y_true, dtype=int),
        np.asarray(y_pred, dtype=int),
        np.asarray(y_prob, dtype=float)
    )


y_true, y_pred, y_prob = predict_model(model, test_loader)

# -------------------- 10. METRICS ---------------------------------

def calculate_metrics(y_true, y_pred, y_prob, n_classes):
    metrics = {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro Precision": precision_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro Recall / Balanced Accuracy": recall_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro F1-score": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Cohen's Kappa": cohen_kappa_score(y_true, y_pred),
        "Matthews Correlation Coefficient": matthews_corrcoef(
            y_true, y_pred
        )
    }

    try:
        if n_classes == 2:
            metrics["ROC-AUC"] = roc_auc_score(y_true, y_prob[:, 1])
            metrics["PR-AUC"] = average_precision_score(
                y_true, y_prob[:, 1]
            )
        else:
            y_true_bin = label_binarize(
                y_true, classes=np.arange(n_classes)
            )
            metrics["ROC-AUC"] = roc_auc_score(
                y_true_bin, y_prob, average="macro", multi_class="ovr"
            )
            metrics["PR-AUC"] = average_precision_score(
                y_true_bin, y_prob, average="macro"
            )
    except (ValueError, IndexError):
        metrics["ROC-AUC"] = np.nan
        metrics["PR-AUC"] = np.nan

    return metrics


test_metrics = calculate_metrics(
    y_true, y_pred, y_prob, num_classes
)

print("\n" + "=" * 68)
print("A1 TEST RESULTS: ConvNeXt-Tiny + ONE-LAYER GCN")
print("=" * 68)

for name, value in test_metrics.items():
    if np.isfinite(value):
        print(f"{name:36s}: {value:.4f} ({value*100:.2f}%)")
    else:
        print(f"{name:36s}: Not available")

# -------------------- 11. BOOTSTRAP 95% CONFIDENCE INTERVALS ------

def bootstrap_confidence_intervals(
    y_true, y_pred, y_prob, n_classes, rounds=2000, seed=42
):
    """
    Percentile bootstrap CIs by resampling test examples.
    These are conditional on this held-out test set and split.
    """
    rng = np.random.default_rng(seed)
    n = len(y_true)

    metric_names = [
        "Accuracy",
        "Macro Precision",
        "Macro Recall / Balanced Accuracy",
        "Macro F1-score",
        "Cohen's Kappa",
        "Matthews Correlation Coefficient",
        "ROC-AUC",
        "PR-AUC"
    ]

    samples = {name: [] for name in metric_names}

    for _ in range(rounds):
        ids = rng.integers(0, n, size=n)
        yt = y_true[ids]
        yp = y_pred[ids]
        prob = y_prob[ids]

        # A bootstrap sample may omit one or more classes.
        current = calculate_metrics(yt, yp, prob, n_classes)

        for name in metric_names:
            value = current.get(name, np.nan)
            if np.isfinite(value):
                samples[name].append(value)

    rows = []
    for name in metric_names:
        estimate = test_metrics.get(name, np.nan)
        values = np.asarray(samples[name], dtype=float)

        if len(values) > 0 and np.isfinite(estimate):
            lower, upper = np.percentile(values, [2.5, 97.5])
        else:
            lower, upper = np.nan, np.nan

        rows.append({
            "Ablation ID": "A1",
            "Model": "ConvNeXt-Tiny + 1-GCN",
            "Metric": name,
            "Estimate": estimate,
            "95% CI Lower": lower,
            "95% CI Upper": upper,
            "Estimate (%)": estimate * 100 if np.isfinite(estimate) else np.nan,
            "95% CI Lower (%)": lower * 100 if np.isfinite(lower) else np.nan,
            "95% CI Upper (%)": upper * 100 if np.isfinite(upper) else np.nan
        })

    return pd.DataFrame(rows)


ci_df = bootstrap_confidence_intervals(
    y_true, y_pred, y_prob,
    n_classes=num_classes,
    rounds=BOOTSTRAP_ROUNDS,
    seed=SEED
)

print("\nMetrics with 95% bootstrap confidence intervals:")
display(ci_df.round(4))

ci_df.to_csv(
    os.path.join(OUTPUT_DIR, "A1_metrics_with_95CI.csv"),
    index=False
)

# -------------------- 12. CONFUSION MATRIX ------------------------

cm = confusion_matrix(
    y_true, y_pred, labels=np.arange(num_classes)
)

fig, ax = plt.subplots(figsize=(max(8, num_classes * 0.65),
                                max(6, num_classes * 0.55)))

im = ax.imshow(cm, interpolation="nearest", cmap="Blues")
ax.set_title("A1: ConvNeXt-Tiny + 1-GCN — Confusion Matrix")
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

ticks = np.arange(num_classes)
ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_xticklabels(class_names, rotation=90, fontsize=8)
ax.set_yticklabels(class_names, fontsize=8)
ax.set_xlabel("Predicted class")
ax.set_ylabel("True class")

threshold = cm.max() / 2 if cm.size and cm.max() > 0 else 0
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(
            j, i, str(cm[i, j]),
            ha="center", va="center",
            color="white" if cm[i, j] > threshold else "black",
            fontsize=7
        )

plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "confusion_matrix.png"),
    dpi=300, bbox_inches="tight"
)
plt.show()

# Save raw confusion matrix
pd.DataFrame(
    cm, index=class_names, columns=class_names
).to_csv(os.path.join(OUTPUT_DIR, "confusion_matrix.csv"))

# -------------------- 13. CLASSIFICATION REPORT -------------------

report = classification_report(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()
print("\nClassification report:")
display(report_df.round(4))

report_df.to_csv(
    os.path.join(OUTPUT_DIR, "classification_report.csv")
)

# -------------------- 14. SAVE PREDICTIONS ------------------------

pred_df = pd.DataFrame({
    "True Label ID": y_true,
    "Predicted Label ID": y_pred,
    "True Class": [class_names[i] for i in y_true],
    "Predicted Class": [class_names[i] for i in y_pred],
    "Correct": y_true == y_pred
})

for i, class_name in enumerate(class_names):
    safe_name = str(class_name).replace(" ", "_")
    pred_df[f"Probability_{safe_name}"] = y_prob[:, i]

pred_df.to_csv(
    os.path.join(OUTPUT_DIR, "test_predictions.csv"),
    index=False
)

# -------------------- 15. SAVE SUMMARY ----------------------------

summary = {
    "Ablation ID": "A1",
    "Model": "ConvNeXt-Tiny + 1-GCN",
    "Dataset Root": ROOT_DIR,
    "Number of Classes": int(num_classes),
    "Class Names": class_names,
    "Total Images": int(len(full_dataset)),
    "Train Images": int(len(train_dataset)),
    "Validation Images": int(len(val_dataset)),
    "Test Images": int(len(test_dataset)),
    "Seed": SEED,
    "Image Size": IMG_SIZE,
    "Batch Size": BATCH_SIZE,
    "GCN Layers": 1,
    "Grid Size": "7x7",
    "Graph Nodes": 49,
    "Graph Edges": "Four-neighbor grid + self-loops",
    "GCN Hidden Channels": GCN_HIDDEN,
    "Backbone": "ConvNeXt-Tiny pretrained, frozen",
    "Trainable Parameters": int(trainable_params),
    "Total Parameters": int(total_params),
    "Best Validation Loss": float(best_val_loss),
    "Training Time Seconds": float(training_seconds),
    "Test Metrics": {
        k: (float(v) if np.isfinite(v) else None)
        for k, v in test_metrics.items()
    }
}

with open(
    os.path.join(OUTPUT_DIR, "A1_experiment_summary.json"),
    "w"
) as f:
    json.dump(summary, f, indent=2)

print("\n" + "=" * 68)
print("ABLATION A1 COMPLETED")
print("=" * 68)
print("Model: ConvNeXt-Tiny + 1-layer GCN")
print("Results directory:", OUTPUT_DIR)
print("\nSaved files:")
for filename in sorted(os.listdir(OUTPUT_DIR)):
    print(" -", filename)
```
